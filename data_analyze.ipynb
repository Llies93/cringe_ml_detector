```python
import pandas as pd

groups = pd.read_csv("data/groups.csv")
departures = pd.read_csv("data/departures.csv")

print("groups:", groups.shape)
print("departures:", departures.shape)

```

```python
# баланс классов — своими глазами, не по цифре из промпта
print(groups['label_a'].value_counts(normalize=True))
print()
print(groups['label_a_free'].value_counts(normalize=True))
print()
print(departures['label_b'].value_counts(normalize=True))
```

```python
# как выглядят ключевые геометрические признаки в целом
geo_cols = ['max_angle_deg', 'spread_km', 'far_km', 'path_km', 'road_savings']
print(groups[geo_cols].describe())
```

```python
# разница в признаках между "опоздавшими" и обычными группами — есть ли вообще видимый сигнал?
print(groups.groupby('label_a')[geo_cols].mean())
```

```python
candidate = ['orders_in_trip', 'max_angle_deg', 'spread_km', 'far_km', 'path_km',
             'path_over_far', 'far_first_ratio', 'road_savings',
             'ready_orders', 'available_couriers', 'load_ratio', 'hour']

print(groups[candidate].isna().sum())
print()
print(groups['orders_in_trip'].value_counts().sort_index())
```
